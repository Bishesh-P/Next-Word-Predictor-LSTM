```python
data = """I love machine learning
I love deep learning
I love artificial intelligence
I like machine learning
I like deep learning
Machine learning is interesting
Machine learning is powerful
Machine learning is useful
Deep learning is powerful
Deep learning is interesting
Deep learning uses neural networks
Artificial intelligence is changing the world
Artificial intelligence is very useful
Neural networks are powerful
Neural networks can learn patterns
Neural networks are used in deep learning
Deep learning can solve complex problems
Machine learning can solve many problems
I am learning machine learning
I am learning deep learning
I am studying artificial intelligence
I am studying neural networks
The model learns from data
The model predicts the next word
The model can learn patterns
The model is trained on data
The neural network learns patterns
The neural network predicts words
Data is important for machine learning
Data is used to train the model
Training the model takes time
Training the neural network takes time
The model learns from examples
The model improves with more data
More data can improve the model
Machine learning models learn from data
Deep learning models use neural networks
Neural networks learn from examples
I want to learn machine learning
I want to learn deep learning
I want to build a neural network
I want to build a prediction model
I want to predict the next word """
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([data])
```

```python
tokenizer.word_index
```

```python
input_sequences = []
for sentence in data.split('\n'):
  token_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1,len(token_sentence)):
    input_sequences.append(token_sentence[:i+1])
```

```python
input_sequences
```

```python
max_len = max([len(i) for i in input_sequences])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

padded_input = pad_sequences(input_sequences, maxlen = max_len,padding = 'pre')
```

```python
padded_input
```

```python
X = padded_input[:,:-1]
print(X)
```

```python
X[0].shape
```

```python
y = padded_input[:,-1]
```

```python
y
```

```python
len(tokenizer.word_index)
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y= to_categorical(y,num_classes = 62)
```

```python
X.shape
```

```python
y
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense , LSTM,Input
```

```python
# model = Sequential()

# model.add(Embedding(62,50,input_length = 6))
# model.add(LSTM(100))
# model.add(Dense(62,activation='softmax'))

# model.summary()
model = Sequential()

model.add(Input(shape=(6,)))
model.add(Embedding(62, 70))
model.add(LSTM(100))
model.add(Dense(62, activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,y,epochs=350)
```

```python
import numpy as np
```

```python
import time
text = "I"

for i in range(10):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=6, padding='pre')
  # predict
  probs = model.predict(padded_token_text, verbose=0)[0]
  pos = np.random.choice(len(probs), p=probs / probs.sum())

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
      time.sleep(2)
```
